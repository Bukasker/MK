```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
import os

# --- KONFIGURACJA ---
# Ścieżka do Twojego folderu na Dysku Google
folder_path = '/content/drive/MyDrive/Folder'

print(f"{'Plik':<10} | {'Rozmiar':<15} | {'Wartość k (90% info)'}")
print("-" * 50)

# --- GŁÓWNA PĘTLA ---
for i in range(1, 14):
    file_name = f"{i}.webp"
    file_path = os.path.join(folder_path, file_name)

    # Sprawdzenie, czy plik istnieje
    if not os.path.exists(file_path):
        print(f"{file_name:<10} | BRAK PLIKU")
        continue

    try:
        # 1. Wczytanie obrazu i konwersja na skalę szarości (L)
        img = Image.open(file_path).convert('L')

        # 2. Konwersja do float64 (wymagane dla poprawnej precyzji SVD)
        A = np.array(img, dtype=np.float64)

        # 3. Dekompozycja SVD
        U, S, Vt = np.linalg.svd(A, full_matrices=False)

        # 4. Obliczenie progu 90% informacji (suma liniowa, bez potęgowania)
        total_info = np.sum(S)
        cumulative_info = np.cumsum(S)

        # 5. Znalezienie najmniejszego k spełniającego warunek >= 90%
        k = np.searchsorted(cumulative_info, 0.90 * total_info) + 1

        # Wyświetlenie wyniku w tabeli
        print(f"{file_name:<10} | {str(A.shape):<15} | {k}")

        # --- OPCJONALNIE: WIZUALIZACJA (odkomentuj poniższe linie, aby zobaczyć efekt) ---
        """
        if i == 1: # Pokaż podgląd tylko dla pierwszego pliku
            Ak = U[:, :k] @ np.diag(S[:k]) @ Vt[:k, :]
            plt.figure(figsize=(10, 5))
            plt.subplot(1, 2, 1)
            plt.imshow(A, cmap='gray')
            plt.title(f"Oryginał {file_name}")
            plt.axis('off')

            plt.subplot(1, 2, 2)
            plt.imshow(Ak, cmap='gray')
            plt.title(f"Skompresowany (k={k})")
            plt.axis('off')
            plt.show()
        """

    except Exception as e:
        print(f"{file_name:<10} | BŁĄD: {e}")

print("\n--- Zakończono analizę wszystkich plików ---")
```
