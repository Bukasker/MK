```python
import numpy as np
import pandas as pd

# 1. Wczytanie danych ze wskazaniem polskiego formatowania
# Używamy sep=';' (częsty separator w polskim CSV) oraz decimal=',' (przecinek dziesiętny)
# Jeśli to nie zadziała, spróbuj zmienić sep na ',' (standardowy przecinek)
df_X = pd.read_csv('War6_X.csv', header=None, sep=';', decimal=',')
df_X_prime = pd.read_csv('War6_Xprime.csv', header=None, sep=';', decimal=',')

# 2. Agresywna konwersja do liczb (wymuszenie)
# Zamienia wszystko co nie jest liczbą na NaN (Not a Number), a potem na zera.
df_X = df_X.apply(pd.to_numeric, errors='coerce').fillna(0)
df_X_prime = df_X_prime.apply(pd.to_numeric, errors='coerce').fillna(0)

# 3. Konwersja do macierzy numpy z jawnym wskazaniem typu float (liczby zmiennoprzecinkowe)
X = df_X.to_numpy(dtype=float)
X_prime = df_X_prime.to_numpy(dtype=float)

# Sprawdzenie typu (powinno wypisać 'float64')
print(f"Typ macierzy X: {X.dtype}")

# ----------------- Dalsza część kodu bez zmian -----------------

# Krok 1: Wyznaczenie SVD
U, Sigma_diag, VT = np.linalg.svd(X, full_matrices=False)

r = len(Sigma_diag)
U_tilde = U[:, :r]
Sigma_tilde = np.diag(Sigma_diag[:r])
V_tilde_star = VT[:r, :]
V_tilde = V_tilde_star.conj().T

# Krok 2: Obliczenie przybliżonej macierzy A
Sigma_inv = np.linalg.inv(Sigma_tilde)
A = X_prime @ V_tilde @ Sigma_inv @ U_tilde.conj().T

print("\nWyznaczona macierz przybliżona A:")
print(A)
```
