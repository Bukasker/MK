```python
import numpy as np
import matplotlib.pyplot as plt

# Funkcja sigmoidalna
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Funkcja Swish
def swish(x):
    return x * sigmoid(x)

# Pochodna Swish
def swish_derivative(x):
    s = sigmoid(x)
    return s + x * s * (1 - s)

# Zakres argumentów
x = np.linspace(-6, 6, 500)

# Obliczenia
y = swish(x)
dy = swish_derivative(x)

# Wykres funkcji
plt.figure(figsize=(8,5))
plt.plot(x, y, label="Swish(x)")
plt.title("Funkcja Swish")
plt.xlabel("x")
plt.ylabel("Swish(x)")
plt.grid(True)
plt.legend()
plt.show()

# Wykres pochodnej
plt.figure(figsize=(8,5))
plt.plot(x, dy, label="Swish'(x)")
plt.title("Pochodna funkcji Swish")
plt.xlabel("x")
plt.ylabel("Swish'(x)")
plt.grid(True)
plt.legend()
plt.show()
```
